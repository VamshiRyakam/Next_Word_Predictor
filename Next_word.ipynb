Data Preprocessing

```python
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from sklearn.model_selection import train_test_split
```

```python
with open('stories.txt','r') as file:
  data =file.read().lower()
```

```python
data
```

Tokenize the text

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])
total_words = len(tokenizer.word_index)+1
total_words
```

```python
tokenizer.word_index
```

Create input sequences

```python
input_sequences = []
for line in text.split('.'):
    token_list = tokenizer.texts_to_sequences([line])[0]
    for i in range(1,len(token_list)):
        n_gram_sequence = token_list[:i+1]
        input_sequences.append(n_gram_sequence)
```

```python
input_sequences
```

padding

```python
max_sequence_len = max([len(i) for i in input_sequences])
max_sequence_len
```

```python
input_sequences = np.array(pad_sequences(input_sequences,maxlen=max_sequence_len,padding='pre'))
input_sequences.shape
```

Train test Split

```python
X,y = input_sequences[:,:-1],input_sequences[:,-1]
```

```python
X
```

```python
y
```

```python
import tensorflow as tf
```

```python
y = tf.keras.utils.to_categorical(y,num_classes=total_words)
```

```python
y
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2)
```

Model Building

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding,LSTM,Dense,Dropout,Input
```

Define the model

```python
model = Sequential()
model.add(Input(shape=(max_sequence_len,)))
model.add(Embedding(input_dim=total_words,output_dim=100))
model.add(LSTM(150,return_sequences=True))
model.add(Dropout(0.2))
model.add(LSTM(100))
model.add(Dense(total_words,activation='softmax'))
```

compile the model

```python
model.compile(loss='categorical_crossentropy',optimizer='adam',metrics=['accuracy'])
model.summary()
```

Train the model

```python
history = model.fit(X_train,y_train,epochs=50,validation_data=(X_test,y_test),verbose=1)
```

```python
def predict_next_word(model,tokenizer,text,max_sequence_len):
  token_list = tokenizer.texts_to_sequences([text])[0]
  if len(token_list) >= max_sequence_len:
    token_list = token_list[-(max_sequence_len):]
  token_list = pad_sequences([token_list], maxlen=max_sequence_len,padding='pre')
  predicted = model.predict(token_list,verbose=0)
  predicted_word_index = np.argmax(predicted,axis=1)
  for word, index in tokenizer.word_index.items():
    if index == predicted_word_index:
      return word
  return None
```

```python
input_text = "Who is enough for her mum who are you going to touch"
print(f"Input text : {input_text}")
max_sequence_len = model.input_shape[1]+1
next_word = predict_next_word(model,tokenizer,input_text,max_sequence_len)
print(f"Next word : {next_word}")
```
